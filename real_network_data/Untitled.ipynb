```python
import pandas as pd
import numpy as np
import networkx as nx
from matplotlib import pyplot as plt
import os
import copy
```

### This network describes the face-to-face behavior of people during the exhibition INFECTIOUS: STAY AWAY in 2009 at the Science Gallery in Dublin. Nodes represent exhibition visitors; edges represent face-to-face contacts that were active for at least 20 seconds.

```python
data = pd.read_csv('./sociopatterns-infectious/data.csv', sep = ' ')
data
```

```python
G = nx.Graph()
for i in range(len(data)):
    G.add_edge(data.iloc[i, 0], data.iloc[i, 1])
plt.figure(figsize = (10, 10))   
nx.draw(G, node_size = 10)
```

```python
G.number_of_nodes()
```

```python
G.number_of_edges()
```
